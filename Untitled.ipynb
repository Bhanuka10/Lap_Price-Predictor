```python
data['CPU_Name']= data['Cpu'].apply(lambda x:"".join(x.split()[0:3]))
```

```python
data['CPU_Name'].value_counts()
```

```python
def set_processor(name):
    if name == 'IntelCorei7' or name == 'IntelCorei5' or name == 'IntelCorei3':
        return name
    
    elif name.split()[0] == 'AMD':
        return 'AMD'
    
    else:
        return 'others'
```

```python
data['CPU_Name']= data['CPU_Name'].apply(set_processor)
```

```python
data['CPU_Name'].value_counts()
```

```python
data['Gpu'].value_counts()
```

```python
data['Gpu_Name']= data['Gpu'].apply(lambda x:"".join(x.split()[0:1]))
```

```python
data['Gpu_Name'].value_counts()
```

```python
data= data[data['Gpu_Name']!='ARM']
```

```python
data['Gpu_Name'].value_counts()
```

```python
data['OpSys'].value_counts()
```

```python
def set_ops(os):
    if os == 'Windows 10' or os == 'Windows 7' or os == 'Windows 10 S':
        return 'Windows'
    
    elif os == "macOS" or os == "Mac OS X":
        return 'macOS'
    
    elif os == "Linux":
        return 'Linux'
    
    else:
        return "others"

data["OpSys"] = data["OpSys"].apply(set_ops)

       
    
```

```python
data['OpSys'].value_counts()
```

```python
data = data.drop(columns=["laptop_ID","Inches","Product","ScreenResolution","Cpu","Gpu"])
```

```python
data
```

```python
data = pd.get_dummies(data)
```

```python
data.head()
```

```python
data.shape
```

```python
x= data.drop("Price_euros", axis=1)
y= data["Price_euros"]
```

```python
pip install scikit-learn
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
X_train.shape
```

```python


# Linear Regression
# Evaluation function
def model_Acc(model):
    model.fit(X_train, y_train)
    acc = model.score(X_test, y_test)
    print(str(model) + " -> " + str(acc))

# Linear Regression
from sklearn.linear_model import LinearRegression
ls = LinearRegression()
model_Acc(ls)

# Decision Tree Regression
from sklearn.tree import DecisionTreeRegressor
dtr = DecisionTreeRegressor()
model_Acc(dtr)

# Random Forest Regression
from sklearn.ensemble import RandomForestRegressor
rfr = RandomForestRegressor()
model_Acc(rfr)

# Gradient Boosting
from sklearn.ensemble import GradientBoostingRegressor
gbr = GradientBoostingRegressor()
model_Acc(gbr)

# XGBoost (optional if installed)
# from xgboost import XGBRegressor
# xgb = XGBRegressor()
# model_Acc(xgb)


```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV

# Define the model
rf = RandomForestRegressor()

# Hyperparameters to tune
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# Grid Search
grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,             # 5-fold cross validation
    scoring='r2',     # evaluation metric for regression
    n_jobs=-1,        # use all CPU cores
    verbose=2
)

# Fit
# Fit
grid_fit = grid_search.fit(X_train, y_train)

# Get best model
best_model = grid_fit.best_estimator_

best_model

# Best parameters

```

```python
best_model.score(X_test, y_test)
```

```python
import pickle

# Suppose your trained model is named 'best_model'
filename = "model.pkl"

with open(filename, 'wb') as file:
    pickle.dump(best_model, file)

print("✅ Model saved successfully!")
```
